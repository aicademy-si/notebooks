Črta iz prvega poskusa je **tangenta**: kaže lokalno smer površja pri izbranem položaju. Njen naklon opisuje **odvod** $h'(x)$. Pozitiven odvod pomeni, da se površje proti desni dviga, negativen pa, da se spušča.


## 2. Spust · Poišči pristajališče v dolini

Zdaj želimo s pomočjo odvoda priti v dolino, kjer je pristajališče. Letalo začne **na pobočju**, preveri trenutni naklon in se premakne v smeri navzdol. Na novi lokaciji postopek ponovi.

Velikost premika uravnavamo s parametrom **α (alfa)**, ki mu pri učenju modelov rečemo tudi **stopnja učenja**. Predstavljaj si ga kot ojačevalnik: večja α pomeni, da trenutni naklon močneje vpliva na premik.

Pri majhni α delamo kratke korake. Pri veliki lahko preskočimo dolino. Pri **α = 0 se sploh ne premaknemo**; pri α = 1 pa od trenutnega položaja odštejemo vrednost odvoda.

Če je $h(x)$ višina površja, naslednji položaj izračunamo tako:

$$x_{\text{naslednji}} = x_{\text{trenutni}} - \alpha\,h'(x_{\text{trenutni}}).$$

Minus izbere lokalno smer navzdol, α pa uravnava velikost premika.


>Primerjaj majhno, zmerno in veliko α. Pred vsakim poskusom izberi nastavitev in pritisni **Samodejni spust**. Za primerjavo ohrani isti začetni položaj; α med spustom ostane enaka.
>
>- **Samodejni spust** izvede celoten poskus. Kratka črta kaže naklon, puščica naslednjo lokacijo, sled pa že obiskane točke.
>- **Začni znova** prekine poskus ter ponastavi položaj, sled in števec.

>**Primerjaj vsaj tri poskuse.** Pri vsakem opazuj število korakov, približevanje ploščadi in morebitno preskakovanje doline.

Zakaj ne izberemo kar največje α?
Zakaj ne izberemo vedno zelo majhne α?

**Kaj pomeni cena koraka?** Drobni premiki pomenijo veliko pristankov in ponovnih vzletov. V zgodbi je to prispodoba za računsko delo: pri učenju modela vsak korak zahteva nov izračun naklona, pogosto iz številnih podatkov. Veliko drobnih izboljšav je zato lahko dražjih kot nekaj dobro izbranih korakov. Prevelik korak pa lahko zapravlja delo s preskakovanjem doline.


## 3. Več dolin · Pristal si. Ampak ali je to najnižja dolina?

Sedaj lahko vidimo širšo pokrajino, želeli bi priti v najnižje možno pristajališče. Postopek ostane enak.

>Najprej pri ponujenem začetnem položaju in α pritisni **Samodejni spust**. Ko se letalo ustavi, poglej tudi druge doline. Ali je našlo najnižje mesto v celotni pokrajini?

Nato ločeno preizkusi vpliv obeh nastavitev:

1. **Ista α, različni začetki — kam pristanemo?** Premikaj **Začetni položaj**, α pa ohrani enako. Za vsak začetek pritisni **Samodejni spust** in primerjaj dosežene doline.
2. **Isti začetek, različne α — ali dosežemo drugo dolino ali samo drugače pridemo do iste?** Izberi en začetni položaj. Spreminjaj samo α in vsakič znova pritisni **Samodejni spust**. Opazuj tudi preskakovanje in morebitne neuspešne pristanke.

### Pošlji 5 letal in preberi tabelo

**Najprej napovej:** ali bodo vsa letala pristala? Če bodo, ali bodo vsa pristala v isti dolini?

Pritisni **Pošlji 5 letal**. Vseh pet uporabi trenutno nastavljeno α, vsako pa začne na drugem mestu. Po koncu poglej tabelo:

- **So vsa pristala?** Preveri stolpec *Izid*. Če katero ni, je bila pri tej nastavitvi njegova pot neuspešna.
- **So pristala enako nizko?** Primerjaj *Doseženo višino*. Pristanek še ne pomeni, da je letalo našlo najnižjo dolino.
- **Kateri poskus je bil najboljši in koliko korakov je potreboval?** Primerjaj višino in število korakov, ne samo tega, ali je letalo pristalo.

Če pri veliki α katero ne pristane, zmanjšaj α in ponovno pošlji pet letal. Preveri, ali uspešnejši pristanki zdaj pomenijo tudi enako končno višino.

### Ali lahko korak prilagodimo obliki doline?

Prvi odvod pove, kako strmo je površje na izbranem mestu. Kaj pa, če bi vedeli tudi, **kako hitro se naklon spreminja**? To opiše **drugi odvod** $h''(x)$. V naši ozki dolini je večji kot v široki.

Poskusimo z njim deliti α:

$$x_{\text{naslednji}} = x_{\text{trenutni}} - \frac{\alpha}{h''(x_{\text{trenutni}})}\,h'(x_{\text{trenutni}}).$$

Večji drugi odvod zmanjša množitelj naklona. Vključi **Upoštevaj drugi odvod** in obdrži α, ki ti je delovala v široki dolini. Preizkusi jo še v ozki. Opazuj tudi prikazano vrednost drugega odvoda in predlagani premik.

Pri naših dolinah je drugi odvod pozitiven. Kdaj lahko takšno pomoč uporabimo tudi drugje, bomo raziskali pozneje.


**Pojasni opažanja:** zakaj je lahko ista α v široki dolini primerna, v ozki pa prevelika? Kako se predlagani premik spremeni, ko vključiš drugi odvod? Ali lahko zdaj obdržiš isto α pri obeh oblikah? **Pri kateri α je upoštevanje drugega odvoda najbolj pomagalo?** Primerjaj uspešnost pristanka in število korakov s pomočjo in brez nje.

Ko upoštevamo drugi odvod, pogosto poskusimo kar **α = 1**. Takrat lahko α iz enačbe izpustimo, ker množimo z 1. Nastavi to vrednost in ponovi samodejni spust v obeh dolinah: pri naših oblikah pridemo do dna v enem koraku. Pri drugih funkcijah je lahko še vedno potrebna manjša α.



## Kaj si pravkar spoznal?

Višino reliefa zapišimo kot funkcijo $h(x)$, položaj letala pa kot $x$.

- **Tangenta** je ravna črta, ki pri izbrani točki pokaže lokalno smer krivulje. Njen naklon je **prvi odvod** $h'(x)$. Z njim si pri vzletu izbiral smer.
- **Gradientni spust** v eni dimenziji ponavlja pravilo
  $$x_{k+1}=x_k-\alpha h'(x_k).$$
  Pozitiven naklon nas pošlje levo, negativen desno. Pri naklonu nič ni predlaganega premika. V večdimenzionalnem primeru vlogo odvoda prevzame gradient.
- **Drugi odvod** $h''(x)$ pove, kako se spreminja naklon. Pri naših dolinah si ga uporabil tudi za prilagoditev koraka obliki doline.
- **Lokalni minimum** je najnižja točka v svoji okolici. **Globalni minimum** je najnižja točka na celotnem obravnavanem območju. Več začetnih poskusov nam pomaga raziskati različne doline.

Pri strojnem učenju je pokrajina prispodoba za **funkcijo napake**. Namesto položaja letala spreminjamo parametre modela, namesto višine pa poskušamo zmanjšati napako. Vsak izračun naklona in preverjanje nove nastavitve nekaj stane, zato je pomembno, kako izbiramo korake.
