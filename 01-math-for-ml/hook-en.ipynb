## 2. Descent · Find a landing spot in the valley

Now we want to use the derivative to reach the valley, where the landing spot is. The plane starts **on a slope**, checks the current slope and moves downhill. At the new position, it repeats the process.

We control the size of the move with a parameter called **α (alpha)**, also known as the **learning rate** when training models. Think of it as an amplifier: a larger α gives the current slope a stronger influence on the move.

With a small α, we take short steps. With a large one, we can overshoot the valley. At **α = 0, we do not move at all**; at α = 1, we subtract the value of the derivative from the current position.

If $h(x)$ is the height of the surface, we calculate the next position as follows:

$$x_{\text{next}} = x_{\text{current}} - \alpha\,h'(x_{\text{current}}).$$

The minus sign selects the local downhill direction, while α controls the size of the move.

**What does a step cost?** Tiny moves mean many landings and take-offs. In the story, this represents computation: when training a model, each step requires a new calculation of the slope, often using a lot of data. Many tiny improvements can therefore cost more than a few well-chosen steps. An overly large step can also waste work by repeatedly overshooting the valley.

### Can we adapt the step to the shape of the valley?

The first derivative tells us how steep the surface is at the chosen position. What if we also knew **how quickly the slope changes**? That is described by the **second derivative** $h''(x)$. In our narrow valley, it is larger than in the wide one.

Let us try dividing α by it:

$$x_{\text{next}} = x_{\text{current}} - \frac{\alpha}{h''(x_{\text{current}})}\,h'(x_{\text{current}}).$$

A larger second derivative reduces the multiplier applied to the slope. Turn on **Use the second derivative** and keep the α that worked in the wide valley. Try it in the narrow one too. Watch the displayed second derivative and the proposed move.

The second derivative is positive in our valleys. We will explore later when this kind of help can also be used elsewhere.

Can you now keep the same α for both shapes? **At which α did using the second derivative help the most?** Compare landing success and the number of steps with and without it.

When we use the second derivative, we often try simply setting **α = 1**. We can then leave α out of the equation, because multiplying by 1 changes nothing. Set this value and repeat the automatic descent in both valleys: for our shapes, we reach the bottom in one step. Other functions may still require a smaller α.



## What have you just discovered?

Let us write the height of the landscape as a function $h(x)$ and the position of the plane as $x$.

- A **tangent** is a straight line that shows the local direction of the curve at a chosen point. Its slope is the **first derivative** $h'(x)$. You used it to choose the take-off direction.
- **Gradient descent** in one dimension repeats the rule
  $$x_{k+1}=x_k-\alpha h'(x_k).$$
  A positive slope sends us left, and a negative one sends us right. At zero slope, no move is proposed. In several dimensions, the gradient takes the role of the derivative.
- The **second derivative** $h''(x)$ tells us how the slope changes. In our valleys, you also used it to adapt the step to the shape of the valley.
- A **local minimum** is the lowest point in its neighbourhood. A **global minimum** is the lowest point over the entire domain being considered. Trying several starting positions helps us explore different valleys.

In machine learning, the landscape represents a **loss function**. Instead of the position of the plane, we change the model parameters; instead of the height, we try to reduce the loss. Each calculation of the slope and evaluation of a new setting has a cost, so the way we choose our steps matters.
