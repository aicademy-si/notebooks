```python
#@title Poženi to celico (klikni ▶ na levi ali pritisni Shift+Enter). Naloži labirint in orodja. Kode ni treba brati.
# Written by Claude

# Every student-visible string of exercise E1 (Slovenian). Keep the {placeholders} and the keys.
# Terms: docs/terminology_sl.md; style: docs/slovene_style.md. Second person singular, present tense (no gendered past forms).
# title/meta/tagline keep the structure the website parses; lines_N are the number words of the block markers.
T = {
    # ---- notebook header (first cell, exactly three lines; a website parses it) ----
    "title": "# Robot v labirintu: od nagrad do strategije",
    "meta": "**Stopnja** 1 · **Predviden čas** približno 45 minut",
    "tagline": "*Formule iz gradiva zapišeš v nekaj vrsticah kode in opazuješ, kako robot z njimi najde najkrajšo pot do cilja.*",
    "setup_title": "Poženi to celico (klikni ▶ na levi ali pritisni Shift+Enter). Naloži labirint in orodja. Kode ni treba brati.",

    # ---- block markers: number words ----
    "lines_1": "1 vrstica",
    "lines_2": "2 vrstici",
    "lines_3": "3 vrstice",
    "lines_4": "4 vrstice",
    "lines_5": "5 vrstic",

    # ---- check_task messages ({n} = task label: 1..6, ★, B1, B2) ----
    "chk_ok": "Naloga {n}: pravilno.",
    "chk_todo": "Naloga {n} še ni rešena: svojo kodo napiši med vrsticama TVOJA KODA in KONEC, nato celico poženi znova.",
    "chk_wrong": "Naloga {n} še ni pravilna. Primer: {case}. Pričakovano: {expected}. Tvoj rezultat: {got}.",
    "chk_error": "Naloga {n}: tvoja koda se je ustavila z napako {error}: {detail}.",
    "chk_unknown": "Naloge z oznako {n} ni.",
    "need_tasks": "Ta celica potrebuje rešene naloge: {tasks}. Najprej jih reši in poženi njihovo preverjanje.",

    # ---- hints printed under chk_wrong ----
    "hint_t1_action": "Namig: akcija mora biti natanko eden od nizov 'up', 'right', 'down', 'left'.",
    "hint_t1_trap": "Namig: pot se konča v pasti (X). Obidi jo.",
    "hint_t1_end": "Namig: pot se ne konča v cilju G. Na sliki zgoraj poglej, kje se robot ustavi.",
    "hint_t3_k1": "Namig: prve nagrade ne znižamo, eksponent k se začne pri 0.",
    "hint_t3_nodisc": "Namig: to je navadna vsota. Vsako nagrado pomnoži z gamma ** k.",
    "hint_t4_sum": "Namig: to je vsota skupnih nagrad. Deli jo s številom epizod, da dobiš povprečje.",
    "hint_t4_nodisc": "Namig: za vsako epizodo uporabi discounted_return (z gamma), ne navadne vsote.",
    "hint_t5_noreward": "Namig: Q(s,a) = R(s,a) + gamma * V(s'). Manjka nagrada za premik: uporabi reward(model, state, action).",
    "hint_t5_sweep": "Namig: sweep mora vrniti nov slovar, v katerem je za vsako stanje največja od njegovih akcijskih vrednosti.",
    "hint_t5_inplace": "Namig: bereš vrednosti, ki so bile v tem preletu že posodobljene. Beri STARE vrednosti in piši v nov slovar.",
    "hint_t5_terminal": "Namig: cilj in pasti nimajo akcij, njihova vrednost ostane 0.",
    "hint_t5_empty_actions": "Pazi: v cilju in pasteh je množica dovoljenih akcij prazna. max() zato nima vrednosti, med katerimi bi izbral največjo.",
    "hint_t6_empty_actions": "Pazi: v cilju in pasteh je množica dovoljenih akcij prazna. Če akcij ni, vrni None, namesto da pokličeš max().",
    "hint_t6_legal": "Namig: vrni eno od dovoljenih akcij v tem stanju (model['transitions'][state]).",
    "hint_t6_reward": "Namig: izberi akcijo z NAJVEČJO akcijsko vrednostjo (nagrada + gamma * vrednost naslednjega stanja).",
    "hint_ch_policy": "Namig: challenge_policy mora biti slovar stanje -> {akcija: verjetnost}. Uporabi policy_from_greedy.",
    "hint_ch_low": "Namig: na challenge_model najprej izvedi iteracijo vrednosti z gamma = GAMMA (manj kot 1), nato iz vrednosti naredi požrešno strategijo.",
    "hint_b1_noslip": "Namig: tvoj rezultat je enak vrednosti brez drsenja. Seštej po vseh izidih, ki jih vrne transition_probs.",
    "hint_b2_alpha": "Namig: Q premakni proti novi oceni samo za delež alpha: Q += alpha * (target - Q).",
    "hint_b2_done": "Namig: ko je done enak True, je epizode konec: nova ocena je samo nagrada, brez vrednosti naslednjega stanja.",
    "hint_b2_greedy": "Namig: pri epsilon = 0 vedno izberi akcijo z največjo vrednostjo Q.",
    "hint_b2_explore": "Namig: pri epsilon = 1 vedno izberi naključno akcijo s seznama actions.",
    "hint_b2_route": "Namig: naučena pot ni najkrajša pot do G. Preveri formulo posodobitve (nova ocena uporabi največjo vrednost Q naslednjega stanja).",

    # ---- environment, errors ----
    "err_maze_shape": "Napaka: labirint mora biti pravokoten in ne sme biti prazen.",
    "err_maze_chars": "Napaka: uporabi samo znake S, G, X, # in .",
    "err_maze_start_goal": "Napaka: labirint mora imeti natanko en S in natanko en G.",
    "err_bad_action": "Neznana akcija {action!r}: uporabi 'up', 'right', 'down' ali 'left'.",

    # ---- plot labels and titles ----
    "lbl_start": "START",
    "lbl_goal": "CILJ",
    "lbl_trap": "PAST",
    "title_maze": "Robot v labirintu",
    "title_walk": "Robot hodi po labirintu",
    "title_walk_values": "Robot hodi po labirintu · številke so vrednosti polj",
    "title_path": "Prehojena pot",
    "title_path_values": "Prehojena pot · spodaj vrednost polja",
    "title_route": "Pot {name}: {moves} premikov",
    "title_sweep1": "Vrednosti po enem preletu",
    "title_vi": "Vrednosti po končani iteraciji vrednosti",
    "title_challenge": "Izziv: nov labirint",
    "title_learned": "Pot, ki se je je robot naučil s Q-učenjem",
    "title_discount": "Koliko šteje nagrada, ki pride čez k korakov",
    "lbl_k": "število korakov k do nagrade",
    "lbl_discount": "teža nagrade $\\gamma^k$",
    "title_hist": "Skupne nagrade naključne strategije",
    "lbl_return": "znižana skupna nagrada ene epizode",
    "lbl_episodes": "število epizod",
    "lbl_mean": "povprečje: {mean:.2f}",
    "title_learning": "Krivulja učenja",
    "lbl_episode": "epizoda",
    "lbl_total_reward": "skupna nagrada v epizodi",
    "lbl_per_episode": "posamezna epizoda",
    "lbl_moving_avg": "povprečje zadnjih {window} epizod",

    # ---- run_robot / run_actions messages ----
    "out_goal": "Cilj dosežen!",
    "out_trap": "Robot je padel v past.",
    "out_stopped": "Cilj ni dosežen: robot se je ustavil po največjem dovoljenem številu korakov.",
    "out_steps_reward": "Koraki: {steps} | Zbrana nagrada: {reward:g}",
    "out_steps": "Koraki: {steps}",
    "out_total": "Skupna nagrada (navadna vsota): {total:g}",
    "run_unfinished": "Akcij je zmanjkalo na polju {cell}. Robot še ni prišel do cilja G (ali v past).",
    "tbl_header": "Prehodi (s = stanje, a = akcija, r = nagrada, s' = novo stanje):\n"
                  "  # | s        | a     |     r | s'",

    # ---- given-cell messages ----
    "t2_total": "Pot {name}: {moves} premikov, navadna vsota nagrad = {total}",
    "t3_return": "gamma = {gamma:g}: pot {name} ima znižano skupno nagrado {value}",
    "t4_mean": "Povprečna znižana skupna nagrada naključne strategije v 300 epizodah: {mean:.3f}",
    "eval_result": "Povprečna znižana skupna nagrada v {episodes} epizodah: {mean:.3f}",
    "eval_no_policy": "Strategije še ni (dokončaj blok zgoraj).",
    "vi_done": "Iteracija vrednosti se je umirila po {sweeps} preletih.",
    "vi_not_settled": "Vrednosti se v {sweeps} preletih niso umirile (je gamma manjši od 1? je nagrada za korak večja od 0?).",
    "vi_bad_sweep": "sweep mora vrniti slovar s številom za vsako stanje. Iteracija vrednosti se je ustavila.",
    "vi_error": "Tvoj sweep se je ustavil z napako {error}: {detail}. Iteracija vrednosti se je ustavila.",
    "cmp_header": "koraki = povprečno število korakov · cilj, past = delež epizod · nagrada = povprečna znižana skupna nagrada\n"
                  + "{:<28} | {:>7} | {:>6} | {:>6} | {:>7}".format("strategija", "koraki", "cilj", "past", "nagrada"),
    "cmp_random": "naključna strategija",
    "cmp_plan": "načrtovanje, gamma = {gamma:g}",
    "cmp_row": "{name:<28} | {steps:7.1f} | {goal:5.0f}% | {trap:5.0f}% | {ret:7.3f}",
    "plan_settings": "Nastavitve: gamma = {gamma:g}, nagrada za korak = {step:g}, nagrada za past = {trap:g}",
    "b1_slip": "Drseča tla, slip = {slip:g}",
    "train_bad_action": "epsilon_greedy mora vrniti eno od akcij 'up', 'right', 'down', 'left'. Učenje se je ustavilo.",

    # ---- quiz ----
    "quiz_correct": "Pravilno!",
    "quiz_wrong": "Ne čisto.",
    "quiz_answers": "Odgovori:",
}

# Maze environment + given helpers for exercise E1. The builder inlines this file into ONE folded setup cell,
# so: stdlib + matplotlib (+ IPython) only, no file/network access, no threads (must run in Pyodide).
import os
import random
import time
from collections import Counter, defaultdict

import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from IPython.display import display


# Tests (and anyone who wants a quiet run) set E1_NO_ANIM=1: no animation, no sleeping.
ANIMATE = os.environ.get("E1_NO_ANIM") != "1"

# ---- maze code below adapted from the teammate's notebook
# 05-reinforcement-learning/5-submodules/10-model-based-methods/2-hook/hook.ipynb (same team), extended with traps,
# model-level rewards and a Gymnasium-style environment.
COLORS = {
    "wall": "#243f49",
    "floor": "#ffffff",
    "goal": "#b9efd5",
    "trap": "#f1b2a8",
    "robot": "#2774d4",
    "arrow": "#203b46",
    "path": "#ca721c",
}

FIGURE_SIZE = (7, 4.5)

DIRECTIONS = {
    "up": (-1, 0),
    "right": (0, 1),
    "down": (1, 0),
    "left": (0, -1),
}
ACTIONS = list(DIRECTIONS)
EVAL_GAMMA = 0.9

# S start, G goal, # wall, . floor, X trap (terminal, optional, any number)
MAZE = [
    "S..#.....",
    ".#.#.#X#.",
    ".#...#...",
    ".###.#.#G",
    ".......#.",
]
# Two routes S -> G that avoid the trap: SHORT is a shortest path (15 moves), LONG goes round (19 moves).
SHORT_ROUTE = ["right", "right", "down", "down", "right", "right", "up", "up",
               "right", "right", "right", "right", "down", "down", "down"]
LONG_ROUTE = ["down", "down", "down", "down", "right", "right", "right", "right",
              "up", "up", "up", "up", "right", "right", "right", "right", "down", "down", "down"]
# Unseen maze for the star challenge: other shape, several traps next to the corridors.
CHALLENGE_MAZE = [
    "S...#....X",
    ".##.#.##..",
    ".#X...#X#.",
    ".#.####.#.",
    ".#...X....",
    "...##.##.#",
    "##X....#.G",
]
# Cliff maze for bonus B1 (slippery floor): the straight route runs along the traps, the safe one is 2 moves longer.
SLIPPERY_MAZE = [
    ".......",
    "S......",
    "XXXX..G",
    "XXXXXXX"
]


def fmt(x, spec="g"):
    """Format a number for messages; a missing answer (None) shows as '?'."""
    if isinstance(x, bool) or not isinstance(x, (int, float)):
        return "?" if x is None else str(x)
    return format(x, spec)


def make_model(maze, step_reward=0, goal_reward=10, trap_reward=-10):
    maze = list(maze)
    if not maze or not maze[0] or any(len(row) != len(maze[0]) for row in maze):
        raise ValueError(T["err_maze_shape"])
    if any(char not in "S.G#X" for row in maze for char in row):
        raise ValueError(T["err_maze_chars"])

    states = [(r, c) for r, row in enumerate(maze)
              for c, char in enumerate(row) if char != "#"]
    starts = [s for s in states if maze[s[0]][s[1]] == "S"]
    goals = [s for s in states if maze[s[0]][s[1]] == "G"]
    traps = [s for s in states if maze[s[0]][s[1]] == "X"]
    if len(starts) != 1 or len(goals) != 1:
        raise ValueError(T["err_maze_start_goal"])
    start, goal = starts[0], goals[0]
    terminal = set(traps) | {goal}
    transitions = {}
    for state in states:
        r, c = state
        transitions[state] = {} if state in terminal else {
            action: (r + dr, c + dc)
            for action, (dr, dc) in DIRECTIONS.items()
            if (r + dr, c + dc) in states
        }
    return {"maze": maze, "states": states, "start": start, "goal": goal, "traps": traps,
            "transitions": transitions, "step_reward": step_reward,
            "goal_reward": goal_reward, "trap_reward": trap_reward}


def step_outcome(model, state, action):
    """(next_state, reward) of one move. Reward is given for the move INTO a cell.
    A move that is not allowed (wall) leaves the robot in place and pays the step reward."""
    next_state = model["transitions"][state].get(action, state)
    if next_state == state:
        return state, model["step_reward"]
    if next_state == model["goal"]:
        return next_state, model["goal_reward"]
    if next_state in model["traps"]:
        return next_state, model["trap_reward"]
    return next_state, model["step_reward"]


def reward(model, state, action):
    return step_outcome(model, state, action)[1]


def transition_probs(model, state, action, slip):
    """List of (probability, next_state, reward). With probability `slip` a random allowed action is executed instead."""
    outcomes = {}

    def add(p, a):
        key = step_outcome(model, state, a)
        outcomes[key] = outcomes.get(key, 0.0) + p

    allowed = list(model["transitions"][state])
    if slip > 0 and allowed:
        add(1 - slip, action)
        for a in allowed:
            add(slip / len(allowed), a)
    else:
        add(1.0, action)
    return [(p, ns, r) for (ns, r), p in outcomes.items()]


def random_policy(model):
    # For the random policy all actions are equally likely
    return {
        state: {action: 1 / len(actions) for action in actions}
        for state, actions in model["transitions"].items()
    }


class MazeEnv:
    """Gymnasium-style environment: reset(seed) -> (obs, info); step(action) -> (obs, reward, terminated, truncated, info).
    obs = (row, col); actions are "up", "right", "down", "left"."""

    def __init__(self, model, slip=0.0, max_steps=200):
        self.model = model
        self.slip = slip
        self.max_steps = max_steps
        self.rng = random.Random()
        self.state = model["start"]
        self.steps = 0
        self.done = False

    def reset(self, seed=None):
        if seed is not None:
            self.rng = random.Random(seed)
        self.state = self.model["start"]
        self.steps = 0
        self.done = False
        return self.state, {}

    def step(self, action):
        if action not in DIRECTIONS:
            raise ValueError(T["err_bad_action"].format(action=action))
        if self.done:
            return self.state, 0.0, True, False, {"outcome": None, "steps": self.steps, "slipped": False}
        model = self.model
        allowed = list(model["transitions"][self.state])
        slipped = False
        if self.slip > 0 and allowed and self.rng.random() < self.slip:
            action, slipped = self.rng.choice(allowed), True
        self.state, r = step_outcome(model, self.state, action)
        self.steps += 1
        terminated = self.state == model["goal"] or self.state in model["traps"]
        truncated = (not terminated) and self.steps >= self.max_steps
        self.done = terminated or truncated
        outcome = ("goal" if self.state == model["goal"] else "trap") if terminated else ("timeout" if truncated else None)
        return self.state, r, terminated, truncated, {"outcome": outcome, "steps": self.steps, "slipped": slipped,
                                                      "action": action}


# ---- episodes -------------------------------------------------------------------------------------------------
def rollout_actions(env, actions):
    """Play a list of actions silently. Returns dict(rewards, path, steps list, outcome, error)."""
    state, _ = env.reset()
    rows, rewards, path, outcome, error = [], [], [state], None, None
    for action in actions:
        try:
            nxt, r, terminated, truncated, info = env.step(action)
        except ValueError:
            error = action
            break
        rows.append((state, action, r, nxt))
        rewards.append(r)
        path.append(nxt)
        state = nxt
        if terminated or truncated:
            outcome = info["outcome"]
            break
    return {"rewards": rewards, "path": path, "rows": rows, "outcome": outcome, "error": error}


def rewards_of(env, actions):
    return rollout_actions(env, actions)["rewards"]


def rollout(env, policy, seed=None):
    """One episode of a policy (dict state -> {action: probability}). Returns dict(rewards, path, steps, outcome)."""
    rng = random.Random(seed)
    state, _ = env.reset(seed=seed)
    rewards, path, outcome = [], [state], "stuck"
    while True:
        probs = policy.get(state)
        if not probs:
            break
        action = rng.choices(list(probs), weights=list(probs.values()))[0]
        state, r, terminated, truncated, info = env.step(action)
        rewards.append(r)
        path.append(state)
        if terminated or truncated:
            outcome = info["outcome"]
            break
    return {"rewards": rewards, "path": path, "steps": len(rewards), "outcome": outcome}


def run_episode(env, policy, seed=None):
    return rollout(env, policy, seed)["rewards"]


def _discounted(rewards, gamma):
    # own copy, so that evaluation never depends on the student's discounted_return
    return sum(gamma ** k * r for k, r in enumerate(rewards))


def rollout_stats(env, policy, episodes=100, gamma=None):
    gamma = EVAL_GAMMA if gamma is None else gamma
    runs = [rollout(env, policy, seed=i) for i in range(episodes)]
    n = len(runs)
    return {
        "mean_return": sum(_discounted(r["rewards"], gamma) for r in runs) / n,
        "mean_steps": sum(r["steps"] for r in runs) / n,
        "goal_share": sum(r["outcome"] == "goal" for r in runs) / n,
        "trap_share": sum(r["outcome"] == "trap" for r in runs) / n,
        "timeout_share": sum(r["outcome"] in ("timeout", "stuck") for r in runs) / n,
    }


def evaluate(env, policy, episodes=100, gamma=None, quiet=False):
    """Mean discounted return of a policy over `episodes` fixed-seed episodes."""
    if not isinstance(policy, dict) or not policy:
        if not quiet:
            print(T["eval_no_policy"])
        return None
    mean = rollout_stats(env, policy, episodes, gamma)["mean_return"]
    if not quiet:
        print(T["eval_result"].format(episodes=episodes, mean=mean))
    return mean


# ---- drawing ---------------------------------------------------------------------------------------------------
def show_maze(model, policy=None, values=None, robot=None, path=None, title=None, show=True):
    fig, ax = plt.subplots(figsize=FIGURE_SIZE)
    maze = model["maze"]
    vmax = max([abs(v) for v in values.values()] + [1e-9]) if values is not None else 1

    for r, row in enumerate(maze):
        for c, char in enumerate(row):
            state = (r, c)
            color = COLORS["wall"] if char == "#" else COLORS["floor"]
            if values is not None and char != "#":
                value = values[state]
                strength = abs(value) / vmax * 0.45
                color = (1 - strength, 1, 1 - strength) if value >= 0 else (1, 1 - strength / 2, 1 - strength)
            if state == model["goal"]:
                color = COLORS["goal"]
            if state in model["traps"]:
                color = COLORS["trap"]
            if state == robot:
                color = COLORS["robot"]
            ax.add_patch(Rectangle((c - .5, r - .5), 1, 1, facecolor=color, edgecolor="#dce6e2", linewidth=1))
            if char == "#":
                continue

            text_color = "white" if state == robot else COLORS["arrow"]
            label = (T["lbl_start"] if state == model["start"] else T["lbl_goal"] if state == model["goal"]
                     else T["lbl_trap"] if state in model["traps"] else "")
            ax.text(c, r - .32, label, ha="center", va="center", fontsize=7, color=text_color)
            if policy is not None:
                for action, probability in policy.get(state, {}).items():
                    if probability > 0:
                        dr, dc = DIRECTIONS[action]
                        ax.arrow(c, r, dc * .24, dr * .24, width=.015, head_width=.12,
                                 length_includes_head=True, color=text_color, zorder=2)
            if values is not None:
                ax.text(c, r + .32, f"{values[state]:.1f}", ha="center", va="center", fontsize=8, color=text_color)

    if path:
        ax.plot([s[1] for s in path], [s[0] for s in path], color=COLORS["path"], alpha=.45, linewidth=2)
    ax.set(xlim=(-.65, len(maze[0]) - .5), ylim=(len(maze) - .5, -.5), aspect="equal")
    ax.axis("off")
    ax.set_title(T["title_maze"] if title is None else title, fontsize=12, pad=14)
    fig.tight_layout()
    if show:
        plt.show()
        plt.close(fig)
        return None
    return fig


def _animate(model, path, policy=None, values=None, title=None, delay=0.03):
    marker = Rectangle((model["start"][1] - .5, model["start"][0] - .5), 1, 1,
                       facecolor=COLORS["robot"], alpha=.85, zorder=1.1)
    fig = show_maze(model, policy, values=values, show=False, title=title)
    fig.axes[0].add_patch(marker)
    plt.close(fig)
    handle = display(fig, display_id=True)
    for r, c in path:
        marker.set_xy((c - .5, r - .5))
        handle.update(fig)
        time.sleep(delay)


def simulate_robot(model, policy, max_steps=120, seed=7, slip=0.0):
    result = rollout(MazeEnv(model, slip=slip, max_steps=max_steps), policy, seed=seed)
    return {"path": result["path"], "steps": result["steps"], "reward": sum(result["rewards"]),
            "reached_goal": result["outcome"] == "goal", "outcome": result["outcome"]}


def _outcome_message(outcome):
    return T["out_goal"] if outcome == "goal" else T["out_trap"] if outcome == "trap" else T["out_stopped"]


def run_robot(model, policy, animate=None, seed=7, max_steps=120, delay=0.03, values=None, slip=0.0):
    animate = ANIMATE if animate is None else animate
    result = simulate_robot(model, policy, max_steps=max_steps, seed=seed, slip=slip)
    if animate:
        _animate(model, result["path"], policy, values, T["title_walk_values"] if values is not None else T["title_walk"],
                 delay)
    else:
        show_maze(model, policy, values=values, robot=result["path"][-1], path=result["path"],
                  title=T["title_path_values"] if values is not None else T["title_path"])
    print(_outcome_message(result["outcome"]))
    print(T["out_steps_reward"].format(steps=result["steps"], reward=result["reward"]))
    return result


def run_actions(env, actions, animate=None):
    """Play a list of actions, print the transitions as a table (s, a, r, s') and return the list of rewards."""
    animate = ANIMATE if animate is None else animate
    played = rollout_actions(env, actions)
    print(T["tbl_header"])
    for i, (s, a, r, s2) in enumerate(played["rows"], 1):
        print(f"{i:>3} | {str(s):<8} | {a:<5} | {r:>5g} | {s2}")
    if played["error"] is not None:
        print(T["err_bad_action"].format(action=played["error"]))
    model = env.model
    if animate:
        _animate(model, played["path"], title=T["title_walk"])
    else:
        show_maze(model, robot=played["path"][-1], path=played["path"], title=T["title_path"])
    if played["outcome"] is None and played["error"] is None:
        print(T["run_unfinished"].format(cell=played["path"][-1]))
    elif played["outcome"] is not None:
        print(_outcome_message(played["outcome"]))
    print(T["out_total"].format(total=sum(played["rewards"])))
    return played["rewards"]


def show_route(env, actions, title=None):
    played = rollout_actions(env, actions)
    show_maze(env.model, robot=played["path"][-1], path=played["path"], title=title)


def plot_discount():
    ks = list(range(31))
    fig, ax = plt.subplots(figsize=(6.5, 3.6))
    for gamma in (1.0, 0.99, 0.9, 0.5):
        ax.plot(ks, [gamma ** k for k in ks], marker="o", markersize=3, label=f"γ = {gamma:g}")
    ax.set_xlabel(T["lbl_k"])
    ax.set_ylabel(T["lbl_discount"])
    ax.set_title(T["title_discount"])
    ax.legend()
    ax.grid(alpha=.3)
    fig.tight_layout()
    plt.show()
    plt.close(fig)


def plot_returns_histogram(returns):
    fig, ax = plt.subplots(figsize=(6.5, 3.6))
    ax.hist(returns, bins=20, color="#2774d4", alpha=.8)
    mean = sum(returns) / len(returns)
    ax.axvline(mean, color="#ca721c", linewidth=2, label=T["lbl_mean"].format(mean=mean))
    ax.set_xlabel(T["lbl_return"])
    ax.set_ylabel(T["lbl_episodes"])
    ax.set_title(T["title_hist"])
    ax.legend()
    fig.tight_layout()
    plt.show()
    plt.close(fig)


def plot_learning_curve(totals, window=20):
    fig, ax = plt.subplots(figsize=(6.5, 3.6))
    ax.plot(totals, color="#b8c4cc", linewidth=.8, label=T["lbl_per_episode"])
    window = max(1, min(window, len(totals)))
    smooth = [sum(totals[max(0, i - window + 1):i + 1]) / len(totals[max(0, i - window + 1):i + 1])
              for i in range(len(totals))]
    ax.plot(smooth, color="#2774d4", linewidth=2, label=T["lbl_moving_avg"].format(window=window))
    ax.set_xlabel(T["lbl_episode"])
    ax.set_ylabel(T["lbl_total_reward"])
    ax.set_title(T["title_learning"])
    ax.legend()
    ax.grid(alpha=.3)
    fig.tight_layout()
    plt.show()
    plt.close(fig)

def train_q_snapshots(env, update_fn, choose_fn, checkpoints=(0, 10, 20, 30, 40, 50),
                      episodes=600, alpha=.5, gamma=.9, epsilon=.2, seed=3):
    rng = random.Random(seed)
    Q = defaultdict(float)
    snapshots = {0: Q.copy()}
    totals = []
    for episode in range(1, episodes + 1):
        state, _ = env.reset(seed=rng.randrange(10 ** 9))
        total = 0.0
        while True:
            action = choose_fn(Q, state, ACTIONS, epsilon, rng)
            next_state, r, terminated, truncated, _ = env.step(action)
            update_fn(Q, state, action, r, next_state, terminated, alpha, gamma)
            total += r
            state = next_state
            if terminated or truncated:
                break
        totals.append(total)
        if episode in checkpoints:
            snapshots[episode] = Q.copy()
    return Q, totals, snapshots


def show_q_snapshots(model, snapshots, choose_fn, max_steps=80, seed=7):
    from matplotlib.patches import Rectangle
    from matplotlib.lines import Line2D

    columns = min(3, len(snapshots))
    rows = (len(snapshots) + columns - 1) // columns
    fig, axes = plt.subplots(rows, columns, figsize=(15, 4.5 * rows), squeeze=False)
    for ax in list(axes.flat)[len(snapshots):]:
        ax.axis("off")
    runs = {}
    for ax, (episode, Q) in zip(axes.flat, snapshots.items()):
        # Preizkus uporablja kopijo tabele: prikaz ne spreminja naučenih ocen.
        shown_Q = Q.copy()
        env = MazeEnv(model, max_steps=max_steps)
        rng = random.Random(seed)
        state, _ = env.reset(seed=seed)
        path = [state]
        outcome = "timeout"
        while True:
            # epsilon = 0: brez raziskovanja; med enakovrednimi akcijami še izbiramo naključno.
            action = choose_fn(shown_Q, state, ACTIONS, 0.0, rng)
            state, _, terminated, truncated, info = env.step(action)
            path.append(state)
            if terminated or truncated:
                outcome = info["outcome"]
                break
        runs[episode] = {"path": path, "outcome": outcome, "steps": len(path) - 1}

        for r, row in enumerate(model["maze"]):
            for c, char in enumerate(row):
                state = (r, c)
                color = (COLORS["wall"] if char == "#" else COLORS["goal"] if char == "G"
                         else COLORS["trap"] if char == "X" else "#f6f7f8")
                ax.add_patch(Rectangle((c - .5, r - .5), 1, 1, facecolor=color,
                                       edgecolor="white", linewidth=1.2))
                if char in "SGX":
                    ax.text(c, r - .31, {"S": "S", "G": "G", "X": "×"}[char],
                            ha="center", va="center", fontsize=9, weight="bold", color=COLORS["arrow"])
                if char in "#GX":
                    continue
                best = max(Q.get((state, a), 0.0) for a in ACTIONS)
                for action in ACTIONS:
                    if abs(Q.get((state, action), 0.0) - best) < 1e-9:
                        dr, dc = DIRECTIONS[action]
                        ax.arrow(c, r, dc * .23, dr * .23, width=.013, head_width=.11,
                                 length_includes_head=True, color="#2774d4", alpha=.75, zorder=2)
        ax.plot([s[1] for s in path], [s[0] for s in path],
                color="#b55a15", linewidth=2, alpha=.85, zorder=3)
        ax.plot(path[-1][1], path[-1][0], "o", color="#b55a15", markersize=4, zorder=4)
        ax.set(xlim=(-.55, len(model["maze"][0]) - .45),
               ylim=(len(model["maze"]) - .45, -.55), aspect="equal")
        ax.axis("off")
        ax.set_title("Pred učenjem" if episode == 0 else f"Po {episode} epizodah",
                     fontsize=13, weight="bold", pad=12)
        status = "Cilj dosežen" if outcome == "goal" else "Cilj ni dosežen"
        detail = " · padec v past" if outcome == "trap" else " · omejitev epizode" if outcome != "goal" else ""
        ax.text(.5, -.06, f"{status}\n{len(path) - 1} korakov{detail}",
                transform=ax.transAxes, ha="center", va="top", fontsize=10,
                color="#247258" if outcome == "goal" else "#a33e38")
    fig.suptitle("Kaj se je robot že naučil? · prikazne epizode brez raziskovanja (ε = 0)", fontsize=14)
    fig.legend(handles=[Line2D([0], [0], color="#2774d4", lw=2, label="Najbolje ocenjene akcije"),
                        Line2D([0], [0], color="#b55a15", lw=2, label="Pot ene epizode")],
               loc="lower center", bbox_to_anchor=(.5, .025), ncol=2, frameon=False)
    fig.subplots_adjust(left=.02, right=.98, top=.9, bottom=.13, wspace=.15, hspace=.6)
    plt.show()
    plt.close(fig)
    return runs





# ---- planning and learning loops (given) -----------------------------------------------------------------------
def _valid_values(model, values):
    return (isinstance(values, dict) and all(
        s in values and isinstance(values[s], (int, float)) and not isinstance(values[s], bool)
        and values[s] == values[s] for s in model["states"]))


def value_iteration(model, gamma, sweep_fn, tolerance=1e-8, max_sweeps=500, quiet=False):
    """Run the student's sweep until the values settle (at most `max_sweeps`). Never raises."""
    values = {state: 0.0 for state in model["states"]}
    for sweeps in range(1, max_sweeps + 1):
        try:
            new_values = sweep_fn(model, values, gamma)
        except Exception as error:  # noqa: BLE001 - students' code: report, never raise
            print(T["vi_error"].format(error=type(error).__name__, detail=error))
            return values
        if not _valid_values(model, new_values):
            print(T["vi_bad_sweep"])
            return values
        change = max(abs(new_values[s] - values[s]) for s in model["states"])
        values = {s: float(new_values[s]) for s in model["states"]}
        if change < tolerance:
            if not quiet:
                print(T["vi_done"].format(sweeps=sweeps))
            return values
    if not quiet:
        print(T["vi_not_settled"].format(sweeps=max_sweeps))
    return values


def train_q_learning(env, update_fn, choose_fn, episodes=300, alpha=0.5, gamma=0.9, epsilon=0.2, seed=0):
    """Q-learning loop that only uses env.reset / env.step. Returns (Q, total reward of each episode).
    Q is a defaultdict(float) keyed by (state, action)."""
    rng = random.Random(seed)
    Q = defaultdict(float)
    totals = []
    for _ in range(episodes):
        state, _ = env.reset(seed=rng.randrange(10 ** 9))
        total = 0.0
        while True:
            action = choose_fn(Q, state, ACTIONS, epsilon, rng)
            if action not in DIRECTIONS:
                print(T["train_bad_action"])
                return Q, totals
            next_state, r, terminated, truncated, _info = env.step(action)
            update_fn(Q, state, action, r, next_state, terminated, alpha, gamma)
            total += r
            state = next_state
            if terminated or truncated:
                break
        totals.append(total)
    return Q, totals


def greedy_route(model, Q, max_len=60):
    """Follow argmax_a Q(s, a) from the start (ties: first action in ACTIONS). Returns (path, outcome)."""
    state, path = model["start"], [model["start"]]
    for _ in range(max_len):
        if not model["transitions"][state]:
            break
        action = max(ACTIONS, key=lambda a: Q[(state, a)])
        state = step_outcome(model, state, action)[0]
        path.append(state)
    outcome = "goal" if state == model["goal"] else "trap" if state in model["traps"] else "lost"
    return path, outcome


def show_learned_route(model, Q):
    path, outcome = greedy_route(model, Q)
    show_maze(model, robot=path[-1], path=path, title=T["title_learned"])
    print(_outcome_message(outcome if outcome != "lost" else "stopped"))
    print(T["out_steps"].format(steps=len(path) - 1))

# check_task(n): never `assert`, never raises to the student. Looks up the student's functions in the notebook
# namespace (globals()), runs them on a few hand-made cases with hard-coded expected values, prints one message from T.
import random
from collections import defaultdict


# Mean discounted return (gamma 0.9, 100 fixed-seed episodes) a policy must reach in the star challenge.
CHALLENGE_THRESHOLD = 1.0  # solution: 1.216 (21 moves), random policy: -0.70


class _Todo(Exception):
    pass


class _Wrong(Exception):
    def __init__(self, case, expected, got, hint=None):
        super().__init__(case)
        self.case, self.expected, self.got, self.hint = case, expected, got, hint


def _fn(name):
    f = globals().get(name)
    if not callable(f):
        raise _Todo()
    return f


def _num(x):
    return isinstance(x, (int, float)) and not isinstance(x, bool)


def _expect(case, got, expected, hints=()):
    """Raise _Todo for None, _Wrong for a different number. `hints`: list of (predicate(got), hint key)."""
    if got is None:
        raise _Todo()
    if _num(got) and abs(got - expected) < 1e-6:
        return
    hint = next((key for pred, key in hints if _num(got) and pred(got)), None)
    raise _Wrong(case, fmt(expected, ".6g"), fmt(got, ".6g") if _num(got) else repr(got), hint)


def _close(a, b):
    return abs(a - b) < 1e-6


# ---- the checks ---------------------------------------------------------------------------------------------------
def _check_1():
    actions = globals().get("my_actions")
    if not isinstance(actions, list) or not actions:
        raise _Todo()
    if any(a not in ACTIONS for a in actions):
        raise _Wrong("my_actions", "up/right/down/left", repr([a for a in actions if a not in ACTIONS][0]), "hint_t1_action")
    played = rollout_actions(MazeEnv(make_model(MAZE)), actions)
    if played["outcome"] == "goal":
        return
    if played["outcome"] == "trap":
        raise _Wrong("my_actions", "G", f"X {played['path'][-1]}", "hint_t1_trap")
    raise _Wrong("my_actions", "G", str(played["path"][-1]), "hint_t1_end")


def _check_2():
    f = _fn("total_reward")
    for rewards, expected in [([0, 0, 10], 10), ([-1, -1, 10], 8), ([], 0)]:
        _expect(f"total_reward({rewards})", f(list(rewards)), expected)


def _check_3():
    f = _fn("discounted_return")
    cases = [([0, 0, 10], 0.9, 8.1), ([1, 1, 1], 1.0, 3), ([5], 0.5, 5), ([1, 2, 3], 0.5, 2.75)]
    for rewards, gamma, expected in cases:
        plain = sum(rewards)
        hints = [(lambda g, e=expected, ga=gamma: ga > 0 and _close(g, e * ga), "hint_t3_k1"),
                 (lambda g, p=plain: _close(g, p), "hint_t3_nodisc")]
        _expect(f"discounted_return({rewards}, {gamma})", f(list(rewards), gamma), expected, hints)


def _check_4():
    f = _fn("mean_return")
    env = MazeEnv(make_model(["S..G"]))
    policy = {(0, 0): {"right": 1.0}, (0, 1): {"right": 1.0}, (0, 2): {"right": 1.0}, (0, 3): {}}
    for gamma, expected in [(0.9, 8.1), (1.0, 10)]:
        hints = [(lambda g, e=expected: _close(g, 4 * e), "hint_t4_sum"),
                 (lambda g: _close(g, 10), "hint_t4_nodisc")]
        _expect(f"mean_return(env, policy, 4, {gamma})", f(env, policy, 4, gamma), expected, hints)


def _check_5():
    av = _fn("action_value")
    corridor = make_model(["S.G"])
    values = {(0, 0): 1.0, (0, 1): 2.0, (0, 2): 0.0}
    cases = [(corridor, (0, 1), "right", 0.9, 10.0), (corridor, (0, 1), "left", 0.9, 0.9),
             (corridor, (0, 0), "right", 0.5, 1.0),
             (make_model(["S.G"], step_reward=-1), (0, 1), "left", 1.0, 0.0),
             (make_model(["SXG"]), (0, 0), "right", 0.9, -10.0)]
    for i, (model, state, action, gamma, expected) in enumerate(cases):
        vals = {s: 0.0 for s in model["states"]} if i == 4 else values
        sa = f"action_value(model, values, {state}, '{action}', {gamma})"
        hints = [(lambda g, m=model, v=vals, s=state, a=action, ga=gamma: _close(g, ga * v[m["transitions"][s][a]]),
                  "hint_t5_noreward")]
        _expect(sa, av(model, vals, state, action, gamma), expected, hints)
    sw = _fn("sweep")
    model = make_model(["G..S"])
    start = {(0, 0): 0.0,(0, 1): 0.0, (0, 2): 0.0, (0, 3): 0.0}
    got = sw(model, dict(start), 0.9)
    if not isinstance(got, dict) or not got:
        raise _Todo()
    expected = {(0, 0): 0.0, (0, 1): 10.0, (0, 2): 0.0, (0, 3): 0.0}
    for state, e in expected.items():
        v = got.get(state)
        if not _num(v) or not _close(v, e):
            hint = ("hint_t5_terminal" if state == (0, 0) else "hint_t5_inplace" if _num(v) and _close(v, 9.0)
                    else "hint_t5_sweep")
            raise _Wrong(f"sweep(model, values, 0.9), stanje {state}", fmt(e), fmt(v) if _num(v) else repr(v), hint)
    got2 = sw(model, got, 0.9)
    for state, e in {(0, 1): 10.0, (0, 2): 9.0, (0, 3): 0.0}.items():
        v = got2.get(state) if isinstance(got2, dict) else None
        if not _num(v) or not _close(v, e):
            raise _Wrong(f"sweep, drugi prelet, stanje {state}", fmt(e), fmt(v) if _num(v) else repr(v), "hint_t5_sweep")


def _check_6():
    ga = _fn("greedy_action")
    model = make_model(["S.G", "..X"])
    values = {(0, 0): 8.1, (0, 1): 9.0, (0, 2): 0.0, (1, 0): 7.29, (1, 1): 8.1, (1, 2): 0.0}
    for state, best, gamma in [((0, 1), {"right"}, 0.9), ((1, 1), {"up"}, 0.9), ((0, 0), {"right"}, 0.9)]:
        got = ga(model, values, state, gamma)
        if got is None:
            raise _Todo()
        if got not in best:
            raise _Wrong(f"greedy_action(model, values, {state}, {gamma})", "/".join(sorted(best)), repr(got),
                         "hint_t6_reward" if got in model["transitions"][state] else "hint_t6_legal")


def _check_challenge():
    policy = globals().get("challenge_policy")
    if not policy:
        raise _Todo()
    if not isinstance(policy, dict):
        raise _Wrong("challenge_policy", "dict", repr(type(policy).__name__), "hint_ch_policy")
    mean = evaluate(MazeEnv(make_model(CHALLENGE_MAZE)), policy, 100, quiet=True)
    if mean < CHALLENGE_THRESHOLD:
        raise _Wrong("evaluate(challenge_env, challenge_policy)", f">= {CHALLENGE_THRESHOLD:g}", fmt(mean, ".3g"),
                     "hint_ch_low")


def _check_b1():
    f = _fn("expected_action_value")
    corridor = make_model(["S.G"])
    values = {(0, 0): 1.0, (0, 1): 2.0, (0, 2): 0.0}
    cases = [((0, 1), "right", 0.9, 0.2, 9.09), ((0, 1), "right", 0.9, 0.0, 10.0), ((0, 0), "right", 0.5, 0.4, 1.0)]
    for state, action, gamma, slip, expected in cases:
        hints = [(lambda g: _close(g, 10.0) and slip > 0, "hint_b1_noslip")]
        _expect(f"expected_action_value(model, values, {state}, '{action}', {gamma}, {slip})",
                f(corridor, values, state, action, gamma, slip), expected, hints)


def _check_b2():
    upd, choose = _fn("q_update"), _fn("epsilon_greedy")
    s, s2 = (0, 0), (0, 1)
    Q = defaultdict(float)
    Q[(s2, "right")] = 10.0
    upd(Q, s, "right", 0.0, s2, False, 0.5, 0.9)
    if Q[(s, "right")] == 0:  # the stub leaves Q untouched
        raise _Todo()
    _expect("q_update(Q, s, 'right', 0, s2, False, alpha=0.5, gamma=0.9)", Q[(s, "right")], 4.5,
            [(lambda g: _close(g, 9.0), "hint_b2_alpha")])
    Q = defaultdict(float)
    Q[(s2, "right")] = 100.0
    upd(Q, s, "right", 10.0, s2, True, 0.5, 0.9)
    _expect("q_update(Q, s, 'right', 10, s2, True, alpha=0.5, gamma=0.9)", Q[(s, "right")], 5.0,
            [(lambda g: g > 5.0, "hint_b2_done")])
    Q = defaultdict(float)
    Q[(s, "down")] = 1.0
    rng = random.Random(1)
    picks = [choose(Q, s, ACTIONS, 0.0, rng) for _ in range(20)]
    if all(p is None for p in picks):
        raise _Todo()
    if set(picks) != {"down"}:
        raise _Wrong("epsilon_greedy(Q, s, ACTIONS, 0.0, rng)", "down", repr(sorted(set(map(str, picks)))), "hint_b2_greedy")
    picks = {choose(Q, s, ACTIONS, 1.0, rng) for _ in range(100)}
    if picks != set(ACTIONS):
        raise _Wrong("epsilon_greedy(Q, s, ACTIONS, 1.0, rng)", "up/right/down/left", repr(sorted(map(str, picks))),
                     "hint_b2_explore")
    model = make_model(MAZE)
    Q, _ = train_q_learning(MazeEnv(model), upd, choose, episodes=B2_EPISODES, seed=B2_SEED)
    path, outcome = greedy_route(model, Q)
    if outcome != "goal" or len(path) - 1 != B2_SHORTEST:
        raise _Wrong(f"train_q_learning(..., episodes={B2_EPISODES})", f"{B2_SHORTEST} premikov do G",
                     f"{len(path) - 1} moves, {outcome}", "hint_b2_route")


B2_EPISODES = 600
B2_SEED = 3
B2_SHORTEST = 15

_CHECKS = {"1": _check_1, "2": _check_2, "3": _check_3, "4": _check_4, "5": _check_5, "6": _check_6,
           "challenge": _check_challenge, "b1": _check_b1, "b2": _check_b2}
_STATUS = {}
_LABELS = {"challenge": "★", "b1": "B1", "b2": "B2"}


def _label(n):
    """Label of a task as the notebook headings show it."""
    return _LABELS.get(str(n).lower(), str(n))


def check_task(n, quiet=False):
    """Check one task. Returns True (correct) or False (not done / wrong); never raises, never asserts."""
    key = str(n).lower()
    check = _CHECKS.get(key)
    if check is None:
        if not quiet:
            print(T["chk_unknown"].format(n=n))
        return False
    try:
        check()
    except _Todo:
        _STATUS[key] = "todo"
        if not quiet:
            print(T["chk_todo"].format(n=_label(n)))
        return False
    except _Wrong as w:
        _STATUS[key] = "wrong"
        if not quiet:
            print(T["chk_wrong"].format(n=_label(n), case=w.case, expected=w.expected, got=w.got))
            if w.hint:
                print(T[w.hint])
        return False
    except Exception as error:  # noqa: BLE001 - students' code: report, never raise
        _STATUS[key] = "error"
        if not quiet:
            print(T["chk_error"].format(n=_label(n), error=type(error).__name__, detail=error))
            if (key in {"5", "6"} and isinstance(error, ValueError)
                    and "max()" in str(error) and "empty" in str(error)):
                print(T[f"hint_t{key}_empty_actions"])
        return False
    _STATUS[key] = "ok"
    if not quiet:
        print(T["chk_ok"].format(n=_label(n)))
    return True


def ready(*tasks):
    """True when the given tasks are solved; otherwise print which ones are missing (used by the given cells)."""
    missing = [str(t) for t in tasks if not check_task(t, quiet=True)]
    if missing:
        print(T["need_tasks"].format(tasks=", ".join(_label(m) for m in missing)))
        return False
    return True

# Given helpers that call the STUDENT's functions (action_value, sweep, greedy_action, discounted_return,
# expected_action_value). In the notebook all cells share one namespace, so the names are resolved at call time.


def greedy_policy(model, values, gamma, q=None):
    """Teammate's greedy policy: every action tied for the best score gets the same probability (ties = random choice)."""
    q = action_value if q is None else q  # noqa: F821 - student's function
    policy = {}
    for state in model["states"]:
        if not model["transitions"][state]:
            policy[state] = {}
            continue
        scores = {action: q(model, values, state, action, gamma) for action in model["transitions"][state]}
        best = max(scores.values())
        best_actions = [a for a, score in scores.items() if abs(score - best) < 1e-8]
        policy[state] = {a: 1 / len(best_actions) for a in best_actions}
    return policy


def policy_from_greedy(model, values, gamma):
    """Deterministic policy that follows the student's greedy_action."""
    return {state: {greedy_action(model, values, state, gamma): 1.0}  # noqa: F821 - student's function
            for state in model["states"] if model["transitions"][state]}


def collect_returns(env, policy, episodes, gamma):
    """Discounted return of many episodes (uses the student's discounted_return); for the histogram."""
    return [discounted_return(run_episode(env, policy, seed=i), gamma) for i in range(episodes)]  # noqa: F821


def slippery_sweep(model, values, gamma, slip):
    """One sweep like the student's, but with expected_action_value (bonus B1)."""
    new_values = {}
    for state in model["states"]:
        actions = model["transitions"][state]
        new_values[state] = max(expected_action_value(model, values, state, a, gamma, slip)  # noqa: F821
                                for a in actions) if actions else 0.0
    return new_values


def compare_planners(maze=None, episodes=50, show=True):
    """Compare all three strategies first, then optionally show example paths. Returns their stats."""
    from html import escape
    from IPython.display import HTML, display

    if not ready(5, 6):
        return None
    model = make_model(MAZE if maze is None else maze)
    env = MazeEnv(model)
    plans = {"random": (T["cmp_random"], random_policy(model), None)}
    for gamma in (1.0, 0.9):
        values = value_iteration(model, gamma, sweep, quiet=True)
        plans[f"gamma_{gamma:g}"] = (T["cmp_plan"].format(gamma=gamma),
                                     greedy_policy(model, values, gamma), values)
    stats = {key: rollout_stats(env, policy, episodes, gamma=EVAL_GAMMA)
             for key, (_, policy, _) in plans.items()}

    headers = ["Strategija", "Povprečno korakov na epizodo", "Dosežen cilj",
               "Padec v past", "Brez zaključka", "Povprečna znižana nagrada"]
    heading = "".join(f'<th style="padding:10px;text-align:left;border-bottom:2px solid #b8c4cc">{h}</th>'
                      for h in headers)
    rows = []
    for key, (label, _, _) in plans.items():
        s = stats[key]
        cells = [escape(label), f'{s["mean_steps"]:.1f}', f'{s["goal_share"]:.0%}',
                 f'{s["trap_share"]:.0%}', f'{s["timeout_share"]:.0%}', f'{s["mean_return"]:.3f}']
        rows.append('<tr>' + ''.join(f'<td style="padding:10px;border-bottom:1px solid #dce6e2">{v}</td>'
                                    for v in cells) + '</tr>')
    print(f"Vsako strategijo preizkusimo v {episodes} epizodah na istem labirintu.")
    display(HTML('<div style="overflow-x:auto"><table style="border-collapse:collapse;width:100%">'
                 f'<thead><tr>{heading}</tr></thead><tbody>{"".join(rows)}</tbody></table></div>'))
    if show:
        for key, (label, policy, values) in plans.items():
            if key != "random":
                print(label)
                run_robot(model, policy, values=values, animate=False, max_steps=env.max_steps)
    return stats


def plan_and_show(maze, gamma=0.9, step_reward=0, trap_reward=-10):
    """Plan with value iteration (the student's sweep), then show values and one robot walk."""
    if not ready(5, 6):
        return None
    model = make_model(maze, step_reward=step_reward, trap_reward=trap_reward)
    values = value_iteration(model, gamma, sweep)  # noqa: F821 - student's function
    policy = greedy_policy(model, values, gamma)
    print(T["plan_settings"].format(gamma=gamma, step=step_reward, trap=trap_reward))
    return run_robot(model, policy, values=values);

import json
QUIZ = json.loads("[{\"id\":\"q01\",\"question\":{\"sl\":\"Robot se uči priti skozi labirint. Kaj od naštetega je del okolja?\",\"en\":\"A robot learns to get through a maze. Which of these is part of the environment?\"},\"options\":{\"sl\":[\"Pravilo, po katerem robot izbira akcije\",\"Zidovi, pasti in nagrade labirinta\",\"Tabela vrednosti, ki jo robot sproti popravlja\"],\"en\":[\"The rule by which the robot chooses actions\",\"The walls, traps and rewards of the maze\",\"The table of values the robot keeps updating\"]},\"answer\":1,\"explain\":{\"sl\":\"Okolje je vse, na kar se agent odziva: labirint, njegova pravila in nagrade. Strategija in tabela vrednosti pripadata agentu.\",\"en\":\"The environment is everything the agent responds to: the maze, its rules and its rewards. The policy and the value table belong to the agent.\"}},{\"id\":\"q02\",\"question\":{\"sl\":\"V čem se razlikujeta nagrada in skupna nagrada?\",\"en\":\"What is the difference between a reward and the return?\"},\"options\":{\"sl\":[\"Nagrada je številka za en korak, skupna nagrada je znižana vsota nagrad od nekega trenutka naprej\",\"Skupna nagrada je največja nagrada v epizodi\",\"Razlike ni, to sta dve imeni za isto\"],\"en\":[\"A reward is a number for one step; the return is the discounted sum of rewards from some moment on\",\"The return is the largest reward of the episode\",\"There is none; they are two names for the same thing\"]},\"answer\":0,\"explain\":{\"sl\":\"Nagrado dobi agent po vsakem koraku. Skupna nagrada sešteje vse nagrade, ki še pridejo, vsako s težo, ki pada z oddaljenostjo.\",\"en\":\"The agent gets a reward after every step. The return adds up all rewards still to come, each with a weight that falls with its distance.\"}},{\"id\":\"q03\",\"question\":{\"sl\":\"Do cilja vodita dve poti, ena s 15 in druga z 19 premiki. Nagrado (+10) da samo cilj. Kaj pove navadna vsota nagrad?\",\"en\":\"Two routes lead to the goal, one with 15 and one with 19 moves. Only the goal gives a reward (+10). What does the plain sum of rewards say?\"},\"options\":{\"sl\":[\"Kratka pot ima večjo vsoto\",\"Obe poti imata enako vsoto\",\"Dolga pot ima večjo vsoto\"],\"en\":[\"The short route has the larger sum\",\"Both routes have the same sum\",\"The long route has the larger sum\"]},\"answer\":1,\"explain\":{\"sl\":\"Vsota je pri obeh 10. Navadna vsota ne upošteva, kdaj nagrada pride, zato kratke poti ne loči od dolge.\",\"en\":\"The sum is 10 for both. The plain sum ignores when a reward arrives, so it cannot tell a short route from a long one.\"}},{\"id\":\"q04\",\"question\":{\"sl\":\"Nagrade v epizodi so [0, 0, 10], faktor znižanja je 0,9. Kolikšna je skupna nagrada?\",\"en\":\"The rewards of an episode are [0, 0, 10] and the discount factor is 0.9. What is the return?\"},\"options\":{\"sl\":[\"10\",\"9\",\"8,1\"],\"en\":[\"10\",\"9\",\"8.1\"]},\"answer\":2,\"explain\":{\"sl\":\"0 + 0,9 · 0 + 0,9² · 10 = 8,1. Tretja nagrada pride dva koraka pozneje, zato jo dvakrat pomnožimo z 0,9.\",\"en\":\"0 + 0.9 · 0 + 0.9² · 10 = 8.1. The third reward comes two steps later, so it is multiplied by 0.9 twice.\"}},{\"id\":\"q05\",\"question\":{\"sl\":\"Kaj se zgodi, če je faktor znižanja zelo majhen, na primer 0,1?\",\"en\":\"What happens when the discount factor is very small, for example 0.1?\"},\"options\":{\"sl\":[\"Agentu so pomembne skoraj samo takojšnje nagrade\",\"Agent bolj upošteva oddaljene nagrade\",\"Vse nagrade postanejo negativne\"],\"en\":[\"Almost only immediate rewards matter to the agent\",\"The agent takes distant rewards into account more\",\"All rewards become negative\"]},\"answer\":0,\"explain\":{\"sl\":\"Teža nagrade čez k korakov je 0,1 na k: že po treh korakih je to samo 0,001. Agent oddaljene nagrade skoraj ne vidi.\",\"en\":\"The weight of a reward k steps away is 0.1 to the power k: after three steps it is only 0.001. The agent hardly sees a distant reward.\"}},{\"id\":\"q06\",\"question\":{\"sl\":\"Kaj pove markovska lastnost?\",\"en\":\"What does the Markov property say?\"},\"options\":{\"sl\":[\"Kaj se zgodi naprej, je odvisno samo od trenutnega stanja, ne od poti, ki je pripeljala do njega\",\"Agent si mora zapomniti vsa pretekla stanja\",\"Vse akcije so enako verjetne\"],\"en\":[\"What happens next depends only on the current state, not on the path that led to it\",\"The agent must remember all past states\",\"All actions are equally likely\"]},\"answer\":0,\"explain\":{\"sl\":\"Trenutno stanje vsebuje vse, kar potrebujemo za napoved naslednjega koraka. Zgodovina ne doda ničesar.\",\"en\":\"The current state holds everything needed to predict the next step. The history adds nothing.\"}},{\"id\":\"q07\",\"question\":{\"sl\":\"Strategija v nekem stanju izbere akcijo »desno« z verjetnostjo 0,7 in akcijo »gor« z verjetnostjo 0,3. Kakšna strategija je to?\",\"en\":\"In some state a policy chooses the action \\\"right\\\" with probability 0.7 and \\\"up\\\" with probability 0.3. What kind of policy is it?\"},\"options\":{\"sl\":[\"Deterministična\",\"Stohastična\",\"Požrešna\"],\"en\":[\"Deterministic\",\"Stochastic\",\"Greedy\"]},\"answer\":1,\"explain\":{\"sl\":\"Stohastična strategija akcijam pripiše verjetnosti. Deterministična bi v istem stanju vedno izbrala isto akcijo.\",\"en\":\"A stochastic policy gives the actions probabilities. A deterministic one would always choose the same action in the same state.\"}},{\"id\":\"q08\",\"question\":{\"sl\":\"V menzi vsak dan naročiš isto jed, ker veš, da je dobra. Kaj počneš?\",\"en\":\"In the canteen you order the same dish every day because you know it is good. What are you doing?\"},\"options\":{\"sl\":[\"Raziskuješ\",\"Izkoriščaš to, kar že veš\",\"Zmanjšuješ težo prihodnjih nagrad\"],\"en\":[\"Exploring\",\"Exploiting what you already know\",\"Discounting\"]},\"answer\":1,\"explain\":{\"sl\":\"Izkoriščanje pomeni izbiro tistega, kar se je doslej izkazalo za najboljše. Kdor nikoli ne poskusi nič novega, lahko spregleda še boljšo jed.\",\"en\":\"Exploitation means choosing what has proved best so far. Whoever never tries anything new may miss an even better dish.\"}},{\"id\":\"q09\",\"question\":{\"sl\":\"Kaj pove akcijska vrednost Q(s, a), česar vrednost stanja V(s) ne pove?\",\"en\":\"What does the action value Q(s, a) tell you that the state value V(s) does not?\"},\"options\":{\"sl\":[\"Kako dobra je posamezna akcija v stanju s\",\"Kolikšna je nagrada za zadnji korak epizode\",\"Koliko korakov je še do cilja\"],\"en\":[\"How good each single action is in state s\",\"The reward of the last step of the episode\",\"How many steps are left to the goal\"]},\"answer\":0,\"explain\":{\"sl\":\"V(s) oceni stanje kot celoto, Q(s, a) pa vsako akcijo v njem posebej. Zato lahko iz vrednosti Q strategijo preberemo neposredno: izberemo akcijo z največjo vrednostjo.\",\"en\":\"V(s) rates the state as a whole, Q(s, a) rates each action in it separately. That is why a policy can be read straight from Q: take the action with the largest value.\"}},{\"id\":\"q10\",\"question\":{\"sl\":\"V čem se Q-učenje razlikuje od iteracije vrednosti?\",\"en\":\"How does Q-learning differ from value iteration?\"},\"options\":{\"sl\":[\"Q-učenje ne uporablja faktorja znižanja\",\"Q-učenje ne potrebuje modela okolja, uči se iz izkušenj\",\"Q-učenje deluje samo v labirintih\"],\"en\":[\"Q-learning does not use a discount factor\",\"Q-learning needs no model of the environment; it learns from experience\",\"Q-learning only works in mazes\"]},\"answer\":1,\"explain\":{\"sl\":\"Iteracija vrednosti načrtuje z znanim modelom: ve, kam vodi vsaka akcija in kakšno nagrado prinese. Q-učenje tega ne ve, zato poskuša in se uči iz prehodov, ki jih doživi.\",\"en\":\"Value iteration plans with a known model: it knows where each action leads and what reward it brings. Q-learning does not know that, so it tries things and learns from the transitions it experiences.\"}}]")

# K quiz: show_quiz(lang="sl") renders the items of quiz.json with ipywidgets radio buttons and instant feedback,
# or prints them as text when ipywidgets cannot be imported. The builder inlines the JSON as QUIZ.



def show_quiz(lang="sl", use_widgets=None):
    """use_widgets: None = use ipywidgets if it can be imported, False = plain text."""
    widgets = None
    if use_widgets is not False:
        try:
            import ipywidgets as widgets
            from IPython.display import display
        except ImportError:
            widgets = None
    if widgets is None:
        _show_quiz_text(lang)
        return

    boxes = []
    for number, item in enumerate(QUIZ, 1):
        feedback = widgets.HTML("")
        radio = widgets.RadioButtons(options=[(text, i) for i, text in enumerate(item["options"][lang])],
                                     value=None, layout=widgets.Layout(width="auto"))

        def on_change(change, item=item, feedback=feedback):
            if change["new"] is None:
                return
            ok = change["new"] == item["answer"]
            head = T["quiz_correct"] if ok else T["quiz_wrong"]
            feedback.value = f"<b>{head}</b> {item['explain'][lang]}"

        radio.observe(on_change, names="value")
        boxes.append(widgets.VBox([widgets.HTML(f"<b>{number}. {item['question'][lang]}</b>"), radio, feedback]))
    display(widgets.VBox(boxes))


def _show_quiz_text(lang):
    for number, item in enumerate(QUIZ, 1):
        print(f"{number}. {item['question'][lang]}")
        for i, text in enumerate(item["options"][lang]):
            print(f"   {'abcdefgh'[i]}) {text}")
    print()
    print(T["quiz_answers"])
    for number, item in enumerate(QUIZ, 1):
        print(f"{number}. {'abcdefgh'[item['answer']]} - {item['explain'][lang]}")
```

## Naloga 5 · Bellmanov korak in iteracija vrednosti

Robot labirint pozna: za vsako polje ve, kam ga pripelje katera akcija in kakšno nagrado dobi. Zato lahko pot načrtuje vnaprej.

Vsakemu stanju pripišemo **vrednost** $V(s)$: skupna nagrada, ki jo robot od tam naprej še lahko pričakuje, če izbira najbolje. Začnemo z ocenami, ki so vse enake 0, in jih postopoma izboljšujemo z dvema formulama.

**Akcijska vrednost.** Koliko je vredna akcija $a$ v stanju $s$? Toliko, kolikor znaša nagrada za premik, plus znižana vrednost polja $s'$, na katero robot pride:

$$Q(s,a) = R(s,a) + \gamma\, V(s')$$

**Nova vrednost stanja** je akcijska vrednost najboljše akcije:

$$V_{\text{nova}}(s) = \max_a Q(s,a)$$

En **prelet** (angl. *sweep*) tako posodobi vsa stanja. Pri tem vedno bere *stare* vrednosti in piše v nov slovar. Cilj in pasti nimajo akcij, njihova vrednost ostane 0.

**Zakaj ima cilj vrednost 0, če prinese nagrado +10?** Nagrado +10 robot dobi **ob vstopu v cilj**. Takrat se epizoda konča, zato iz cilja ne pričakuje več prihodnjih nagrad: njegova vrednost je 0. Enako velja za past: kazen −10 dobi ob vstopu, vrednost samega terminalnega stanja pa je 0. Nagrada pripada prehodu, vrednost stanja pa opisuje nagrade od tega stanja naprej.
