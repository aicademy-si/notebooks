## Preberi

**Spodbujevalno učenje** (angl. *reinforcement learning*) je način, kako se računalnik uči s poskušanjem. Nihče mu ne pokaže prave rešitve. Dobi samo **nagrado** (ali kazen) za to, kar je naredil.

**Pravila igre.** Pacman se sprehaja po labirintu, polnem pik, in jih poskuša pojesti čim več. Ko poje vse, zmaga. Po labirintu pa se potika tudi duh: če ujame Pacmana, je igre konec. Velike pike (kapsule) duha za kratek čas prestrašijo, da se upočasni in beži, Pacman pa ga lahko takrat poje in dobi dodatne točke. Igra traja največ 400 korakov.

Naša igra je poenostavljena: labirint je majhen, duh pa samo en. Tudi ta ni posebno pameten: včasih gre naravnost proti Pacmanu (približno 4-krat od 10), drugače pa zavije v naključno smer. Ko je prestrašen, večinoma beži.

Pojme agent, okolje, stanje, akcija in nagrada podrobno razložimo v besedilu, ki ga prebereš po tem zvezku. Tu samo to, kaj Pacman ve o igri (njegovo **stanje**). Za vsak možen premik pogleda polje, na katero bi prišel, in oceni:

- kako daleč je najbližja pika in ali jo s tem premikom poje,
- kako daleč je duh in ali je tik ob njem,
- kako daleč je kapsula,
- kako daleč je prestrašeni duh.

Na začetku Pacman ne ve ničesar, zato se premika naključno. Po vsakem koraku pogleda, kakšno nagrado je dobil,
in malo popravi svojo oceno, kako dobra je bila ta akcija v takem stanju. Tej oceni pravimo **Q-vrednost**,
takemu učenju pa **Q-učenje** (angl. *Q-learning*).

Pacman si ne zapomni vsakega polja v labirintu. Raje ocenjuje splošne lastnosti stanja, na primer »pika je blizu«
(dobro) ali »duh je tik ob meni« (slabo). Zato lahko naučeno uporabi tudi v labirintu, ki ga še ni videl.

Po nekaj sto igrah so njegove odločitve že pametne. Cilja igre pa Pacman ne pozna. Ve le, katere akcije mu prinesejo več nagrad.
