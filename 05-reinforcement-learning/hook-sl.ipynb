## Preberi

**Spodbujevalno učenje** (angl. *reinforcement learning*) je način, kako se računalnik uči s poskušanjem. Nihče mu ne pokaže prave rešitve. Dobi samo **nagrado** (ali kazen) za to, kar je naredil.

Definirajmo igro Pacmana:

- **Agent** je Pacman.
- **Okolje** je labirint s pikami (hrano), kapsulami in duhovi.
- **Stanje** je to, kar Pacman vidi (hrana, duhovi, zidovi).
- **Akcija** je premik gor, dol, levo ali desno, lahko pa Pacman tudi ostane na mestu.
- **Nagrada** (angl. *reward*) je število točk glede na rezultat akcije v določenem stanju. Lahko je pozitivna (nagrada) ali negativna (kazen).

Na začetku Pacman ne ve ničesar, zato se premika naključno. Po vsakem koraku pogleda, kakšno nagrado je dobil,
in malo popravi svojo oceno, kako dobra je bila ta akcija v takem stanju. Tej oceni pravimo **Q-vrednost**,
takemu učenju pa **Q-učenje** (angl. *Q-learning*).

Pacman si ne zapomni vsakega polja v labirintu. Raje ocenjuje splošne lastnosti stanja, na primer »pika je blizu«
(dobro) ali »duh je tik ob meni« (slabo). Zato lahko naučeno uporabi tudi v labirintu, ki ga še ni videl.

Po nekaj sto igrah so njegove odločitve že pametne. Cilja igre pa Pacman ne pozna. Ve le, katere akcije mu prinesejo več nagrad.
